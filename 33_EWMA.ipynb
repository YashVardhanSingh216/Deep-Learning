```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('DailyDelhiClimateTest.csv')
df = df.drop(columns = ['humidity', 'wind_speed', 'meanpressure'])
df.head()
```

```python
plt.scatter(df['date'], df['meantemp'], c = 'red')
```

```python
x1 = df['meantemp'].ewm(alpha = 0.9).mean()
x1
```

```python
df['ewma'] = x1
df.head()
```

```python
plt.scatter(df['date'], df['meantemp'], c = 'red')
plt.plot(df['date'], df['ewma'], c = 'blue', label = 'alpha = 0.9')
plt.legend()
plt.show()
```

```python
x2 = df['meantemp'].ewm(alpha = 0.8).mean()
plt.scatter(df['date'], df['meantemp'], c = 'red')
plt.plot(df['date'],x2, c = 'blue', label = 'alpha = 0.8')
plt.legend()
plt.show()
```
