## ***Demo of Perceptron on Iris Dataset\***

```python
df = pd.read_csv("iris.csv")
df.head()
```

```python
df = df.drop(columns = ['sepal_width', 'petal_width'] )
df.head()
```

```python
df['species'].unique()
```

```python
sns.scatterplot(data = df, x = 'sepal_length', y = 'petal_length', hue = 'species')
```

```python
df.shape[0]
```

```python
x = []
y = []
for i in range(df.shape[0]):
    if(df['species'][i] != 'virginica'):
        x.append([df['sepal_length'][i], df['petal_length'][i]])
        if(df['species'][i] == 'setosa'):
            y.append(0)
        else:
            y.append(1)
```

```python
# 0 -> Setosa
# 1 -> Versicolor

x = pd.DataFrame(x, columns = ['sepal_length', 'petal_length'])
y = pd.DataFrame(y, columns = ['species'])
```

```python
x.head()
```

```python
new_df = pd.concat([x,y], axis = 1)
new_df.sample(5)
```

```python
sns.scatterplot(data = new_df, x = 'sepal_length', y = 'petal_length', hue = 'species')
```

```python
x = new_df.iloc[:,:-1];
y = new_df['species']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
p.fit(x_train,y_train)
```

```python
# w1 , w2
p.coef_
```

```python
# b
p.intercept_
```

z = w1*x1 + w2*x2 + b

z = (-3.9)*x1 + (8.7)*x2 + (-2)






```python
p.score(x_test,y_test)
```

```python
p.predict([[5,5]])
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x_train.values, y_train.values, clf=p)
```

```python
plot_decision_regions(x_test.values, y_test.values, clf=p)
```
