```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['input1']=[1,1,0,0]
or_data['input2']=[1,0,1,0]
or_data['output']=[1,1,1,0]
```

```python
or_data
```

```python
sns.scatterplot(data = or_data, x = 'input1', y = 'input2', hue = 'output', s = 200)
```

```python
from sklearn.linear_model import Perceptron
p1 = Perceptron() # or
p1.fit(or_data[['input1','input2']],or_data['output'])
```

```python
p1.coef_
```

```python
p1.intercept_
```

```python
m = -p1.coef_[0][0]/p1.coef_[0][1]
c = -p1.intercept_[0]/p1.coef_[0][1]
print(m,c) # y = -x + 0.5
```

```python
x_input = np.linspace(-1,1,5)
y_input = m*x_input + c
```

```python
sns.scatterplot(data = or_data, x = 'input1', y = 'input2', hue = 'output', s = 200)
plt.plot(x_input,y_input, c = 'red')
plt.show()
```

```python
and_data['input1']=[1,1,0,0]
and_data['input2']=[1,0,1,0]
and_data['output']=[1,0,0,0]
```

```python
and_data
```

```python
sns.scatterplot(data = and_data, x = 'input1', y = 'input2', hue = 'output', s = 200)
```

```python
p2 = Perceptron()
p2.fit(and_data[['input1','input2']],and_data['output'])
```

```python
p2.coef_
```

```python
p2.intercept_
```

```python
m = -p2.coef_[0][0]/p2.coef_[0][1]
c = -p2.intercept_[0]/p2.coef_[0][1]
print(m,c) # m = -x + 1
```

```python
x_input = np.linspace(-1,1,5)
y_input = m*x_input + c
```

```python
sns.scatterplot(data = and_data, x = 'input1', y = 'input2', hue = 'output', s = 200)
plt.plot(x_input,y_input, c = 'red')
plt.show()
```

```python
xor_data['input1']=[1,1,0,0]
xor_data['input2']=[1,0,1,0]
xor_data['output']=[0,1,1,0]
```

```python
xor_data
```

```python
sns.scatterplot(data = xor_data, x = 'input1', y = 'input2', hue = 'output', s = 200)
```

```python
p3 = Perceptron() # xor
p3.fit(xor_data[['input1','input2']],xor_data['output'])
```

```python
plot_decision_regions(xor_data[['input1','input2']].values, xor_data['output'].values, clf=p3)
plt.show()
```
